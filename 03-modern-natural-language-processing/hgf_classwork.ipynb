### Zero shot classification

```python
model = pipeline("zero-shot-classification")
```

```python
model("This is a course on Huggingface", candidate_labels=["politics", "education", "sports", "business"])
```

```python
model("Infosys stocks rose by 2% yesterday", candidate_labels=["politics", "education", "sports", "business"])
```

### Using Specific Model

```python
from transformers import pipeline
model = pipeline("text-generation", model="HuggingFaceTB/SmolLM2-360M-Instruct")
```

```python
text = model("Gravity is a force that ..." )
```

```python
print(text[0]["generated_text"])
```

### Wrong Use Case

```python
model('classify the statement into ["sports", "politics", "business", "education"]: This is an AI course')
```

### Image Classification

```python
url = r"https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/pipeline-cat-chonk.jpeg"
```

```python
image_classifier = pipeline(
    task="image-classification",
    model="google/vit-base-patch16-224"
)
```

```python
image_classifier(url)
```

```python
image_classifier(r"https://images.pexels.com/photos/112460/pexels-photo-112460.jpeg?cs=srgb&dl=automobile-automotive-car-112460.jpg&fm=jpg")
```
